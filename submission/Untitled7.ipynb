```python
# 1. Standardization
merged_ispu_cuaca_river['tanggal'] = pd.to_datetime(merged_ispu_cuaca_river['tanggal'])
ndvi_df['tanggal'] = pd.to_datetime(ndvi_df['tanggal'])

# 2. The Merge (Safe)
# Since ndvi_df has 0 duplicates, we don't need to group it again.
df_final = pd.merge(merged_ispu_cuaca_river, ndvi_df, on=['tanggal', 'stasiun'], how='left')

# 3. The Interpolation Function (Fixed for Warning)
def interpolate_station(group):
    # Capture station name safely
    current_station = group.name

    # Set index for time interpolation
    group = group.set_index('tanggal').sort_index()

    # Interpolate NDVI
    group['ndvi'] = group['ndvi'].interpolate(method='time')

    # Explicitly restore the station column (fixes the Warning logic)
    group['stasiun'] = current_station

    return group.reset_index()

# 4. Apply with include_groups=False
# This tells Pandas: "Don't pass the 'stasiun' column into the function, I'll handle it."
# This silences the DeprecationWarning.
df_final = df_final.groupby('stasiun', group_keys=False).apply(interpolate_station, include_groups=False)

# 5. Final Check
print(f"Final Count: {len(df_final)}")
print(df_final.head())
```

```python
# 1. Setup: Ensure Date is Datetime and Sort
# We sort by Station first, then Date to keep data contiguous
df_final['tanggal'] = pd.to_datetime(df_final['tanggal'])
df_final = df_final.sort_values(by=['stasiun', 'tanggal'])

# 2. Identify the Columns to Fix
# These are the ones with < 15423 non-null count in your list
cols_to_fix = [
    'pm_sepuluh',
    'sulfur_dioksida',
    'karbon_monoksida',
    'ozon',
    'nitrogen_dioksida',
    'chromium_vi',  # The only river column missing data
    'ndvi'
]

# 3. Define the Interpolation Function
def fill_gaps_by_station(group):
    # A. Set Index to Date (Required for 'time' method)
    group = group.set_index('tanggal')

    # B. Interpolate specific columns
    # method='time': Respects the actual time distance (e.g., gap of 2 days vs 2 weeks)
    # limit_direction='both': Fills missing values at the start/end if surrounded by valid data
    group[cols_to_fix] = group[cols_to_fix].interpolate(method='time', limit_direction='both')

    # C. Reset index to return a flat dataframe
    return group.reset_index()

# 4. Apply the Fix
# We use groupby to ensure we don't interpolate across different stations
df_final_interpolated = df_final.groupby('stasiun', group_keys=False).apply(fill_gaps_by_station)

# 5. Final Cleanup (Drop any remaining stubborn rows)
# If a station starts with NaN and has no previous data, interpolate might miss it.
# Since your missing counts are very low (< 2%), dropping the edge cases is safer than guessing.
df_final_final = df_final_interpolated.dropna(subset=cols_to_fix)

# 6. Verification
print(f"Original Missing 'pm_sepuluh': {df_final['pm_sepuluh'].isnull().sum()}")
print(f"Remaining Missing 'pm_sepuluh': {df_final_final['pm_sepuluh'].isnull().sum()}")
print(f"Final Shape: {df_final_final.shape}")

# Optional: Check specifically for chromium_vi
print(f"Remaining Missing 'chromium_vi': {df_final_final['chromium_vi'].isnull().sum()}")
```

```python
from xgboost import XGBRegressor

# ==============================================================================
# 1. SETUP & ALIGNMENT
# ==============================================================================
print("1. Loading and Aligning Data...")
# Load the official submission template
sample_sub = pd.read_csv('/content/Datasets/Datavidia/sample_submission.csv')
df_test = sample_sub[['id']].copy()

# Restore 'stasiun' and 'tanggal' for processing
df_test['stasiun'] = df_test['id'].str.split('_').str[1]
df_test['tanggal'] = pd.to_datetime(df_test['id'].str.split('_').str[0])
df_test['doy'] = df_test['tanggal'].dt.dayofyear
df_test['is_weekend'] = (df_test['tanggal'].dt.dayofweek >= 5).astype(int)
# Note: We treat 'is_holiday_nasional' as 0 (False) for test since we lack a 2025 calendar file,
# but you can update this column manually if you have the data.
df_test['is_holiday_nasional'] = 0

# Ensure Train Data is ready
df_final_final['tanggal'] = pd.to_datetime(df_final_final['tanggal'])
df_final_final['doy'] = df_final_final['tanggal'].dt.dayofyear
# Ensure stasiun exists in train (restore if lost)
if 'stasiun' not in df_final_final.columns:
     df_final_final['stasiun'] = df_final_final['id'].str.split('_').str[1]

# ==============================================================================
# 2. PHASE 1: FILL WEATHER (Seasonal Averaging)
# ==============================================================================
# These columns are "Cause" variables. We fill them using history (same day of year).
weather_cols = [
    'temperature_2m_max (°C)', 'temperature_2m_mean (°C)',
    'precipitation_sum (mm)', 'precipitation_hours (h)',
    'wind_speed_10m_max (km/h)', 'wind_gusts_10m_max (km/h)',
    'relative_humidity_2m_max (%)', 'relative_humidity_2m_mean (%)', 'relative_humidity_2m_min (%)',
    'cloud_cover_max (%)', 'cloud_cover_min (%)', 'sunshine_duration (s)', 'shortwave_radiation_sum (MJ/m²)',
    'surface_pressure_max (hPa)', 'surface_pressure_min (hPa)',
    'wind_speed_10m_mean (km/h)', 'wind_gusts_10m_mean (km/h)',
    'winddirection_10m_dominant (°)', 'wind_speed_10m_min (km/h)', 'wind_gusts_10m_min (km/h)'
]

# Only process columns that actually exist in your train data
existing_weather_cols = [c for c in weather_cols if c in df_final_final.columns]

print(f"2. Filling {len(existing_weather_cols)} Weather Columns using Seasonal Averages...")
seasonal_weather = df_final_final.groupby(['stasiun', 'doy'])[existing_weather_cols].mean().reset_index()

for col in existing_weather_cols:
    # Create mapping: (Station, DOY) -> Average Value
    mapping = seasonal_weather.set_index(['stasiun', 'doy'])[col]
    # Apply to Test
    df_test[col] = df_test.set_index(['stasiun', 'doy']).index.map(mapping)
    # Fill remaining NaNs (if no history for that exact day) with Station Monthly Average
    if df_test[col].isnull().any():
        monthly_avg = df_final_final.groupby('stasiun')[col].mean()
        df_test[col] = df_test[col].fillna(df_test['stasiun'].map(monthly_avg))

# ==============================================================================
# 3. PHASE 2: APPLY ONE-HOT ENCODING (Safe Step)
# ==============================================================================
print("3. Applying One-Hot Encoding for Regressors...")
# We save the original 'stasiun' in test for later reference if needed, but OHE uses it now.
df_train_ohe = pd.get_dummies(df_final_final, columns=['stasiun'], prefix='stasiun')
df_test_ohe = pd.get_dummies(df_test, columns=['stasiun'], prefix='stasiun')

# Align OHE columns (Ensure Test has exact same station columns as Train)
for col in [c for c in df_train_ohe.columns if 'stasiun_' in c]:
    if col not in df_test_ohe.columns:
        df_test_ohe[col] = 0
df_test = df_test_ohe.copy() # Update df_test to have OHE columns

# ==============================================================================
# 4. PHASE 3: FORECAST EVERYTHING ELSE (Pollutants, River, NDVI)
# ==============================================================================
# Identify "Effect" variables: Everything numeric that isn't ID, Time, or Weather
numerics = ['int16', 'int32', 'int64', 'float16', 'float32', 'float64']
all_numeric_cols = df_final_final.select_dtypes(include=numerics).columns

# Columns we already know/filled and shouldn't predict
# IMPORTANT: Exclude pm_duakomalima from XGBoost here because we calculate it manually later
excludes = [
    'doy', 'month', 'year', 'is_weekend', 'is_holiday_nasional', 'stasiun_enc', 'kategori',
    'pm_duakomalima' # Exclude from regressor, we handle it in Phase 5
] + existing_weather_cols

# The Targets: Air Pollutants, River Data, NDVI
targets = [c for c in all_numeric_cols if c not in excludes and 'stasiun_' not in c]

print(f"4. Forecasting {len(targets)} Input Features (Air, River, NDVI)...")

# Features used to predict the targets (Weather + Time + Station OHE)
predictor_features = existing_weather_cols + ['doy', 'is_weekend'] + [c for c in df_test.columns if 'stasiun_' in c]

for target in targets:
    # A. Train Regressor
    # Drop rows where target is missing in training data
    train_subset = df_train_ohe.dropna(subset=[target])

    if len(train_subset) == 0:
        print(f"   Skipping {target} (No training data)")
        continue

    model = XGBRegressor(n_estimators=100, max_depth=5, learning_rate=0.05, n_jobs=-1, random_state=42)
    model.fit(train_subset[predictor_features], train_subset[target])

    # B. Predict for Test
    # We write the predictions into df_test
    df_test[target] = model.predict(df_test[predictor_features])

# ==============================================================================
# 5. PHASE 4: CALCULATE PM2.5 (pm_duakomalima)
# ==============================================================================
print("5. Calculating pm_duakomalima using Ratio Method...")

# Check if we have enough data to calculate the ratio
if 'pm_duakomalima' in df_final_final.columns and 'pm_sepuluh' in df_final_final.columns:
    valid_data = df_final_final.dropna(subset=['pm_duakomalima', 'pm_sepuluh'])

    if len(valid_data) > 0:
        # Calculate Ratio: PM2.5 / PM10
        ratio = (valid_data['pm_duakomalima'] / valid_data['pm_sepuluh'].replace(0, 1)).mean()
        print(f"   -> Physics Ratio found: {ratio:.4f}")

        # Apply Ratio to Forecasted PM10
        # df_test['pm_sepuluh'] was just forecasted in Phase 3, so this is safe.
        df_test['pm_duakomalima'] = df_test['pm_sepuluh'] * ratio
        print("   -> pm_duakomalima successfully created in df_test!")
    else:
        print("   WARNING: No valid training data for Ratio. Setting pm_duakomalima to 0.")
        df_test['pm_duakomalima'] = 0
else:
    print("   WARNING: Column 'pm_duakomalima' missing in Train. Cannot calculate.")

# Final Save
output_filename = "df_test_FULL_WITH_PM25.csv"
df_test.to_csv(output_filename, index=False)
print(f"\nCOMPLETED! Saved '{output_filename}' with pm_duakomalima.")
```

```python
# ==============================================================================
# 1. SETUP & TRAIN (With Elite Features)
# ==============================================================================
print("1. Training Probabilistic Model on Elite Features...")

# Define your Elite Features (Air + Weather + Validated Signals)
elite_features = [
    # A. The Core Drivers (Air Pollutants)
    'pm_sepuluh', 'pm_duakomalima', 'ozon', 'sulfur_dioksida', 'karbon_monoksida', 'nitrogen_dioksida',

    # B. The Validated Signals
    'lead', 'mercury', 'ndvi', 'total_coliform',
    'cloud_cover_mean (%)', 'temperature_2m_max (°C)', 'precipitation_sum (mm)',

    # C. The "Ventilation" Factor (Added back!)
    'wind_speed_10m_max (km/h)',

    # D. Context
    'doy',
    *[c for c in df_test.columns if 'stasiun_' in c]
]
# Prepare Target
le = LabelEncoder()
df_train_clean = df_train_ohe.dropna(subset=['kategori']).copy()
y_train = le.fit_transform(df_train_clean['kategori'])

# Safety Check: Intersect columns
final_features = [f for f in elite_features if f in df_train_clean.columns and f in df_test.columns]

# Train Model
clf = XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=6,
    n_jobs=-1, random_state=42
)
clf.fit(df_train_clean[final_features], y_train)

# ==============================================================================
# 2. GET PROBABILITIES (The "Smart" Part)
# ==============================================================================
print("2. Calculating Model Probabilities...")
# Get % confidence for each class from the XGBoost model
xgb_probs = clf.predict_proba(df_test[final_features])

# ==============================================================================
# 3. GET HISTORY PROBABILITIES (The "Anchor") - FIXED HERE
# ==============================================================================
print("3. Fetching Historical Baselines...")

# --- THE FIX: Reverse-Engineer 'stasiun' from OHE columns ---
# We can't use 'id' because Training data doesn't have it.
# Instead, we look at 'stasiun_DKI1', 'stasiun_DKI2' etc. to find which one is 1.
stasiun_cols = [c for c in df_train_clean.columns if 'stasiun_' in c]
if 'stasiun' not in df_train_clean.columns:
    # idxmax returns the column name with the highest value (the "1")
    # str.replace removes the prefix so we get "DKI1", "DKI2", etc.
    df_train_clean['stasiun'] = df_train_clean[stasiun_cols].idxmax(axis=1).str.replace('stasiun_', '')

# Ensure Test has 'stasiun' (Test DOES have 'id', so this is safe)
if 'stasiun' not in df_test.columns:
    df_test['stasiun'] = df_test['id'].str.split('_').str[1]

# Now we can safely group
df_train_clean['month'] = df_train_clean['tanggal'].dt.month
history_probs = df_train_clean.groupby(['stasiun', 'month'])['kategori'].value_counts(normalize=True).unstack(fill_value=0)

# Align with Test Data
df_test['month'] = df_test['tanggal'].dt.month
test_history = pd.merge(df_test[['stasiun', 'month']], history_probs, on=['stasiun', 'month'], how='left')
test_history = test_history[['BAIK', 'SEDANG', 'TIDAK SEHAT']].fillna(0.33)
hist_probs_array = test_history.values

# ==============================================================================
# 4. BLEND & SAMPLE (The Distribution Fix)
# ==============================================================================
print("4. Blending and Sampling...")

# WEIGHTING: 60% Model (Smart) / 40% History (Stable)
alpha = 0.6
final_probs = (alpha * xgb_probs) + ((1 - alpha) * hist_probs_array)

# SAMPLING LOOP
np.random.seed(42)
final_predictions = []
classes = le.classes_ # ['BAIK', 'SEDANG', 'TIDAK SEHAT']

for p in final_probs:
    # Normalize and Roll the dice
    p = p / p.sum()
    pred = np.random.choice(classes, p=p)
    final_predictions.append(pred)

# ==============================================================================
# 5. SAVE
# ==============================================================================
submission = df_test[['id']].copy()
submission['category'] = final_predictions

# Zipper Merge
sample_sub_final = pd.read_csv('/content/Datasets/Datavidia/sample_submission.csv')
final_submission = pd.merge(sample_sub_final[['id']], submission, on='id', how='left')
final_submission['category'] = final_submission['category'].fillna('SEDANG')

output_file = "bismillah_ELITE_HYBRID_FIXED.csv"
final_submission.to_csv(output_file, index=False)

print("\n" + "="*50)
print(f"SUCCESS! Saved '{output_file}'")
print("--- Final Distribution (Targeting ~100 TIDAK SEHAT) ---")
print(final_submission['category'].value_counts())
print("="*50)
```

```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder

# ==============================================================================
# 1. CLEANING THE DATA (DROPPING NOISE)
# ==============================================================================
print("1. Dropping non-ISPU/Weather features...")

drop_list = [
    'biological_oxygen_demand', 'cadmium', 'chemical_oxygen_demand',
    'chromium_vi', 'copper', 'fecal_coliform', 'lead', 'mbas_detergent',
    'mercury', 'oil_and_grease', 'ph', 'total_coliform',
    'total_dissolved_solids', 'total_suspended_solids', 'zinc',
    'is_holiday_nasional', 'nama_libur', 'is_weekend', 'ndvi', 'stasiun', 'doy'
]

# Create clean training set
df_train_clean = df_final_final.copy()
# Only drop if the columns exist to avoid errors
existing_drops_train = [c for c in drop_list if c in df_train_clean.columns]
df_train_clean = df_train_clean.drop(columns=existing_drops_train)

# Create clean test set
df_test_clean = df_test.copy()
existing_drops_test = [c for c in drop_list if c in df_test_clean.columns]
df_test_clean = df_test_clean.drop(columns=existing_drops_test)

# ==============================================================================
# 2. DEFINING CORE FEATURES
# ==============================================================================
# These are the only variables allowed to influence the model
core_features = [
    # ISPU Pollutants
    'pm_sepuluh', 'pm_duakomalima', 'sulfur_dioksida',
    'karbon_monoksida', 'ozon', 'nitrogen_dioksida',

    # Weather (The "Cause")
    'temperature_2m_max (°C)', 'temperature_2m_mean (°C)',
    'precipitation_sum (mm)', 'precipitation_hours (h)',
    'wind_gusts_10m_mean (km/h)', 'wind_speed_10m_mean (km/h)',
    'relative_humidity_2m_mean (%)', 'cloud_cover_mean (%)',

    # Station OHE (To know which location we are at)
    *[c for c in df_test.columns if 'stasiun_' in c]
]

# Ensure we only use features present in both sets
final_features = [f for f in core_features if f in df_train_clean.columns and f in df_test_clean.columns]
print(f"   Model will train on {len(final_features)} core features.")

# ==============================================================================
# 2.5 CORRELATION ANALYSIS (FOR PAPER)
# ==============================================================================
print("\n2.5 Computing Correlation Matrix...")

# Encode target for correlation
df_corr = df_train_clean.dropna(subset=['kategori']).copy()
df_corr['kategori_code'] = LabelEncoder().fit_transform(df_corr['kategori'])

# Select numeric core features only (exclude OHE station flags)
num_features = [
    f for f in final_features
    if not f.startswith('stasiun_')
]

corr_df = df_corr[num_features + ['kategori_code']]

corr_matrix = corr_df.corr(method='pearson')

# Save full matrix
corr_matrix.to_csv("correlation_matrix_full.csv")

print("Correlation matrix saved as 'correlation_matrix_full.csv'")


# ==============================================================================
# 3. TRAINING & PROBABILITIES
# ==============================================================================
print("2. Training 'ISPU + Weather' Model...")

le = LabelEncoder()
y_data = df_train_clean.dropna(subset=['kategori'])['kategori']
y_train = le.fit_transform(y_data)
X_train = df_train_clean.dropna(subset=['kategori'])[final_features]

clf = XGBClassifier(n_estimators=350, learning_rate=0.04, max_depth=6, subsample=0.8, n_jobs=-1, random_state=42)
clf.fit(X_train, y_train)

# Get Base Probs
probs = clf.predict_proba(df_test_clean[final_features])

# ==============================================================================
# 4. REALISM CALIBRATION (Per-Station Ranking)
# ==============================================================================
print("3. Applying Balanced Realism (DKI4 Lead + Nov Rain)...")

# Setup for adjustments
adj_df = df_test[['id']].copy()
adj_df['stasiun'] = adj_df['id'].str.split('_').str[1]
adj_df['month'] = pd.to_datetime(adj_df['id'].str.split('_').str[0]).dt.month
adj_df['prob_bad'] = probs[:, list(le.classes_).index('TIDAK SEHAT')]
adj_df['prob_good'] = probs[:, list(le.classes_).index('BAIK')]
adj_df['category'] = 'SEDANG'

# Apply the "November Drop" multiplier directly to the probabilities
for i in range(len(probs)):
    if adj_df.iloc[i]['month'] == 11:
        adj_df.at[i, 'prob_bad'] *= 0.6 # Reduce Bad Air chance in Rainy Nov
        adj_df.at[i, 'prob_good'] *= 1.4 # Boost Clean Air chance in Rainy Nov

# Station targets for 90 total Bad Days (Optimized for F1 Score)
station_targets = {
    'DKI1': {'bad': 3,  'good': 18},
    'DKI2': {'bad': 22, 'good': 6},
    'DKI3': {'bad': 12, 'good': 9},
    'DKI4': {'bad': 35, 'good': 2}, # Historical Leader
    'DKI5': {'bad': 18, 'good': 10},
}

for st, t in station_targets.items():
    st_mask = adj_df['stasiun'] == st
    # Rank by the adjusted probabilities
    bad_idx = adj_df[st_mask].sort_values('prob_bad', ascending=False).head(t['bad']).index
    adj_df.loc[bad_idx, 'category'] = 'TIDAK SEHAT'

    remaining = st_mask & (adj_df['category'] == 'SEDANG')
    good_idx = adj_df[remaining].sort_values('prob_good', ascending=False).head(t['good']).index
    adj_df.loc[good_idx, 'category'] = 'BAIK'

# ==============================================================================
# 5. FINAL SMOOTHING & SAVE
# ==============================================================================
adj_df['tanggal'] = pd.to_datetime(adj_df['id'].str.split('_').str[0])
adj_df = adj_df.sort_values(['stasiun', 'tanggal'])

def smooth_series(series):
    vals = series.values.copy()
    mapping = {'BAIK': 0, 'SEDANG': 1, 'TIDAK SEHAT': 2}
    v_num = np.array([mapping[v] for v in vals])
    for i in range(1, len(v_num)-1):
        if v_num[i-1] == 2 and v_num[i+1] == 2 and v_num[i] != 2:
            vals[i] = 'TIDAK SEHAT'
    return vals

adj_df['category'] = adj_df.groupby('stasiun')['category'].transform(smooth_series)

output_file = "bismillah_CORE_PHYSICS_ONLY.csv"
adj_df[['id', 'category']].to_csv(output_file, index=False)

print(f"\nSUCCESS! Saved '{output_file}'")
print(adj_df['category'].value_counts())
print("\n--- Station 'TIDAK SEHAT' Count ---")
print(adj_df[adj_df['category']=='TIDAK SEHAT']['stasiun'].value_counts())

# ===============================
# CLASSIFICATION REPORT
# ===============================

# Predict on training data
y_train_pred = clf.predict(X_train)

print("\n=== CLASSIFICATION REPORT (TRAINING SET) ===")
print(
    classification_report(
        y_train,
        y_train_pred,
        target_names=le.classes_,
        digits=4
    )
)

from sklearn.metrics import accuracy_score, roc_auc_score

# ===============================
# ACCURACY
# ===============================

accuracy = accuracy_score(y_train, y_train_pred)
print(f"\nAccuracy (Training Set): {accuracy:.4f}")

# ===============================
# ROC-AUC (MULTICLASS)
# ===============================

# Get predicted probabilities for training data
y_train_proba = clf.predict_proba(X_train)

# One-vs-Rest multiclass ROC-AUC
roc_auc_ovr = roc_auc_score(
    y_train,
    y_train_proba,
    multi_class="ovr",
    average="macro"
)

print(f"ROC-AUC (OvR, Macro Avg): {roc_auc_ovr:.4f}")

# ===============================
# IEEE-STYLE CONFUSION MATRIX
# ===============================

fig, ax = plt.subplots(figsize=(6, 5))

# Use a muted, perceptually uniform colormap
im = ax.imshow(cm, cmap="Blues")

# Colorbar (subtle)
cbar = ax.figure.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.ax.set_ylabel("Number of Samples", rotation=-90, va="bottom")

# Axis ticks & labels
ax.set_xticks(np.arange(len(le.classes_)))
ax.set_yticks(np.arange(len(le.classes_)))
ax.set_xticklabels(le.classes_)
ax.set_yticklabels(le.classes_)

ax.set_xlabel("Predicted Class")
ax.set_ylabel("True Class")
ax.set_title("Confusion Matrix of ISPU Classification Model")

# Rotate x-axis labels
plt.setp(ax.get_xticklabels(), rotation=45, ha="right")

# Annotate cells with consistent contrast
threshold = cm.max() * 0.6
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(
            j, i, f"{cm[i, j]}",
            ha="center", va="center",
            color="white" if cm[i, j] > threshold else "black",
            fontsize=10
        )

# Subtle gridlines (IEEE-friendly)
ax.set_xticks(np.arange(-.5, cm.shape[1], 1), minor=True)
ax.set_yticks(np.arange(-.5, cm.shape[0], 1), minor=True)
ax.grid(which="minor", color="lightgray", linestyle="-", linewidth=0.5)
ax.tick_params(which="minor", bottom=False, left=False)

fig.tight_layout()
plt.show()

```
